```python
import random
import time
import requests


URL = "http://localhost:1702/api/churn/?user_id=5"


def generate_data():
    return {
        "floor": random.randint(1, 20),
        "kitchen_area": random.uniform(5, 30),
        "living_area": random.uniform(10, 100),
        "rooms": random.randint(0, 4),
        "is_apartment": random.choice([True, False]),
        "studio": random.choice([True, False]),
        "total_area": random.uniform(20, 150),
        "build_year": random.randint(1900, 2029),
        "building_type_int": random.randint(1, 10),
        "latitude": random.uniform(55.5, 55.9),
        "longitude": random.uniform(37.3, 37.9),
        "ceiling_height": random.uniform(2.0, 4.0),
        "flats_count": random.randint(1, 200),
        "floors_total": random.randint(1, 30),
        "has_elevator": random.choice([True, False]),
    }


for i in range(1, 1010):
    print(f"Starting request {i}")

    data = generate_data()

    try:
        response = requests.post(
            URL,
            json=data,
            timeout=30
        )

        print(
            f"Request {i} -> HTTP {response.status_code}"
        )

    except requests.RequestException as e:
        print(f"Request {i} -> ERROR: {e}")

    # 200 мс до следующего запроса
    time.sleep(0.001)


print("Load test completed.")
```
